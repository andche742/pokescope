## Pokemon base stats

```python
pokemon = stats_scraper.parse_entries(stats_scraper.load_raw(PROJECT_ROOT / "pokemon.json"))
pokemon[pokemon["types"].apply(len) == 1]
```

# Pokemon CP calculation


```python
for pkm in ["Lickitung"]:
    level = 50
    r = cpm[cpm["level"] == level].iloc[0]
    multiplier = r["value"]

    row = pokemon[pokemon["name"] == pkm].iloc[0]
    a_iv, d_iv, s_iv = 15, 0, 0
    a, d, s = row["attack"] + a_iv, row["defense"] + d_iv, row["hp"] + s_iv,
    cp = max(10, math.floor((a ) * (d )**0.5 * (s )**0.5 * multiplier**2/ 10))



    print(f"{pkm} @ Level {level}: CP {cp}")
```
